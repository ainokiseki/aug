```python
from queue import Queue
```

```python
f=open('graph1.txt','r')
g={}
for i in f.readlines():
    j=i.replace('\n','')
    j=j.split(' ')
    if j[0] not in g:
        g[j[0]]=[]
    if j[1] not in g[j[0]]:
        g[j[0]].append(j[1])
    if j[1] not in g:
        g[j[1]]=[]
    if j[0] not in g[j[1]]:
        g[j[1]].append(j[0])
```

```python
vis=set()
res=[]
for i in g:
    fres=[]
    if i in vis:
        continue
    q=Queue()
    q.put(i)
    vis.add(i)
    while not q.empty():
        x=q.get()
        fres.append(x)

        for j in g[x]:
            if j not in vis:
                vis.add(j)
                q.put(j)
    res.append(fres)
```

```python
[len(i) for i in res]
```

```python
w=open('graph1k.txt','w')
ct=0
for i in res[0]:
    for j in g[i]:
        if int(i)<int(j):
            w.write(i+'\t'+j+'\n')
            ct+=1
w.close()
```
