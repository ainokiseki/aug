```python
g=open('graph8.txt','w')
d={}
idx=1
with open('graph8k.txt','r') as f:
    
    for i in f.readlines():
        j=i.replace('\n','')
        j=i.split('\t')
        if j[0] not in d:
            d[j[0]]=idx
            idx+=1
        if j[1] not in d:
            d[j[1]]=idx
            idx+=1
        g.write
```
